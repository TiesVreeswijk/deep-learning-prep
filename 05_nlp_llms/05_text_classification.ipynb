## 1. Baseline: TF-IDF + Logistic Regression

```python
import time
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

t0 = time.time()
vec = TfidfVectorizer(max_features=20_000, ngram_range=(1, 2))
X_train = vec.fit_transform(ds["train"]["text"])
X_test = vec.transform(ds["test"]["text"])
clf = LogisticRegression(max_iter=1000).fit(X_train, ds["train"]["label"])
baseline_acc = accuracy_score(ds["test"]["label"], clf.predict(X_test))
print(f"TF-IDF + LR: {baseline_acc:.3f} in {time.time() - t0:.1f}s")
```
